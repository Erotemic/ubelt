# Ubelt's cmd method

```python
import ubelt as ub
```

```python
info = ub.cmd(['python', '--version'], shell=False)
type(info)
print(info.keys())
print(ub.urepr(info, sort=0))
```

```python
info = ub.cmd(['python', '--verssion'])
print(info['ret'])
```

```python
print(ub.cmd.__doc__)
```

```python
ub.cmd('echo hello world && ping localhost -c 3', verbose=3, shell=True)
```

```python
try:
    ub.cmd('ping localhost', verbose=3, shell=True, timeout=3)
except Exception as ex:
    print(ex)
```

```python
try:
    ub.cmd('ping localhost', verbose=3, shell=False, timeout=3)
except Exception as ex:
    print(ex)
```
